## **a. Por medio de la librería ‘Pandas’, lea la base de datos, asigne nombres para las columnas, realice una imputación para la variable ‘Bare Nuclei’ por medio de la mediana y cambie las categorías para la variable ‘Class’ de 2 o 4 por 0 o 1.**

### **Lectura de base de datos y asignación de nombres para las columnas**

```python
import pandas as pd
import numpy as np
```

```python
url = 'https://archive.ics.uci.edu/ml/machine-learning-databases/breast-cancer-wisconsin/breast-cancer-wisconsin.data'
columnas = ['Sample_code_number', 'Clump_thickness', 'Uniformity_of_cell_size', 'Uniformity_of_cell_shape', 'Marginal_adhesion',
            'Single_epithelial_cell_size', 'Bare_nuclei', 'Bland_chromatin', 'Normal_nucleoli', 'Mitoses', 'Class']
df = pd.read_csv(url, names=columnas)
df.head()
```

### **Realizar una imputación para la variable ‘Bare Nuclei’ por medio de la mediana**

```python
# Verificar el tipo de datos de Bare_nuclei
df.dtypes
```

```python
# Convertir la columna a numérico
df['Bare_nuclei'] = pd.to_numeric(df['Bare_nuclei'], errors='coerce')
```

```python
df.dtypes
```

```python
# Imputación de la variable 'Bare_Nuclei' con la mediana
median = df['Bare_nuclei'].median()
df['Bare_nuclei'].fillna(median, inplace=True)
```

```python
df.isnull().sum()
```

### **Cambio las categorías para la variable ‘Class’ de 2 o 4 por 0 o 1.**

```python
# Cambio de categorías para la variable 'Class': 2 o 4 por 0 o 1
df['Class'] = np.where(df['Class'] == 2, 0, df['Class'])
df['Class'] = np.where(df['Class'] == 4, 1, df['Class'])
```

```python
df.head()
```

### **Detectar de valores atípicos univariados por medio del método del rango intercuartílico con 3 de longitud a la derecha y 3 a la izquierda, y elimínarlos.**

```python
from sklearn.preprocessing import RobustScaler
from scipy.stats import chi2
from sklearn.covariance import MinCovDet
```

```python
# Detección de valores atípicos univariados (método del rango intercuartílico)
Q1 = df['Class'].quantile(0.25)
Q3 = df['Class'].quantile(0.75)
IQR = Q3 - Q1
limite_inferior = Q1 - 3 * IQR
print('Valor atipicos del limite inferior',limite_inferior)
limite_superior = Q3 + 3 * IQR
print('Valor atipicos del limite superior',limite_superior)
```

```python
df['Class']
```

```python
# Eliminación de valores atípicos univariados
df_clean = df[(df['Class'] >= limite_inferior) & (df['Class'] <= limite_superior)]
```

### **Detección de valores atípicos multivariados por medio de las distancias de Mahalanobis y elimine aquellos valores que superen el valor de 30.**

```python
from scipy.spatial.distance import mahalanobis
```

```python
# Seleccionamos las columnas numéricas (ajusta según tu dataset)
cols = ['Clump_thickness', 'Uniformity_of_cell_size']

X = df[cols]
```

```python
# Media y matriz de covarianza inversa
mean = X.mean().values
cov = np.cov(X.values.T)  # Usamos X.T porque debe ser una matriz de 2x2
inv_cov = np.linalg.inv(cov)

# Calcular la distancia de Mahalanobis para cada fila
dist_maha = X.apply(lambda row: mahalanobis(row.values, mean, inv_cov), axis=1)

# Agregar la distancia como una nueva columna
df['mahalanobis_distance'] = dist_maha
```

```python
df['mahalanobis_distance']
```

```python
# Filtrar los datos que tienen distancia menor o igual a 30
df_clean_maha = df[df['mahalanobis_distance'] <= 30].copy()
```
